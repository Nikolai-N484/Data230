# Question 3

***True*** or False: The square root of 441 is an integer.

```python
# How to calculate the square root

num = float(input('Enter a number: '))

num_sqrt = num ** 0.5
print('The square root of %0.3f is %0.3f'%(num ,num_sqrt))
```

Notice in the code that we used the exponential operator ** to raise it to 1/2 power which is another way of saying square root.

# Question 5

***True*** or False: A 100,000 loan at 6% interest compounded monthly after 7 years would be worth more than $150,000.

Edit the code below to enter the values (after the equals sign) for each variable in the calculation. For example, if the principal of the loan was $150,000, you would enter p = 150000 # principal, starting amount

```python
from math import exp

p = 100000  # principal, starting amount
r = 0.06  # interest rate, by year
t = 7  # time, number of years
n = 12  # number of times compounded

Question_5 = p * (1 + (r/n))**(n * t)

print("%.2f" % Question_5)
```
